```python
my_density = gaussian_kde(real_estate_agency_df['SalePrice'])
# my_density = gaussian_kde(real_estate_agency_df['SalePrice'], bw_method = 5)
```

```python
x = np.linspace(min(real_estate_agency_df['SalePrice']), max(real_estate_agency_df['SalePrice']), 1000)
plt.plot(x, my_density(x), 'g')
plt.hist(real_estate_agency_df['SalePrice'], density = True, alpha = 0.3)
plt.show()
```

```python
plt.plot(x, my_density(x), 'blue')
```

```python
real_estate_agency_df.groupby('MS Zoning')['SalePrice'].plot.hist(density = True, alpha = 0.6)
plt.legend()
```
